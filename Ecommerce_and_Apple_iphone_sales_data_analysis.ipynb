# **TOP HIGHEST RATED IPHONES ON FLIPKART IN INDIA**

```python
highest_rated=data.sort_values(by = ["Star Rating"],ascending = False)

highest_rated = highest_rated.head(10)
```

```python
print(highest_rated['Product Name'])
```

```python
print(highest_rated)
```

# HOW MANY RATINGS DO THE HIGHEST RATED IPHONES ON FLIPKART HAVE?

```python
# Step 1: Group by product name and sum the number of ratingss
iphone_Starrating = data.groupby("Product Name")["Star Rating"].sum().reset_index()

# Step 2: Sort in descending order (highest ratings first)
iphone_Starrating = iphone_Starrating.sort_values(by="Star Rating", ascending=False)

# Step 3: Select only the top 1
top_Starrating = iphone_Starrating.head(1)
print(top_Starrating)
```

```python
iphones = highest_rated['Product Name'].value_counts()
label = iphones.index
counts = highest_rated["Number Of Ratings"]
figure = px.bar(highest_rated, x=label, y=counts, title ="number of ratings of highest rated iphones")
figure.show()
```

# **WHICH IPHONE HAS THE HIGHEST NUMBER OF REVIEWS ON FLIPKART?**

```python
data.head()
```

```python
iphones = highest_rated['Product Name'].value_counts()
label = iphones.index
counts = highest_rated["Number Of Reviews"]
figure = px.bar(highest_rated, x=label, y=counts, title ="number of reviews of highest rated phones")
figure.show()
```

# **Iphone with highest number of reviews on FlipKart"**

```python
iphones = data['Product Name'].value_counts()
label = iphones.index
counts = data["Number Of Reviews"]
figure = px.bar(iphones, x=label, y=counts, title ="Iphone with highest number of reviews on FlipKart")
figure.show()
```

```python

# Step 1: Group by product name and sum the number of reviews
iphone_reviews = data.groupby("Product Name")["Number Of Reviews"].sum().reset_index()

# Step 2: Sort in descending order (highest reviews first)
iphone_reviews = iphone_reviews.sort_values(by="Number Of Reviews", ascending=False)

# Step 3: Select only the top 1
top_10_reviews = iphone_reviews.head(1)
print(top_10_reviews)
```

# **what is the relationship between sales price of iphone and number of ratings on flipkart**

```python
data.head()
```

```python
figure = px.scatter(data_frame = data, x='Number Of Ratings', y='Sale Price', size='Discount Percentage', trendline= 'ols', title='relationship between sales price and number of ratings')
figure.show()
```

There is a negative realtionship between the sale price of iphones and the number of ratings. Lower cost iphones have higher ratings.

# **what is the relationship between discount price of iphone and number of ratings on flipkart**

```python
fig=px.scatter(data_frame=data, x= 'Number Of Ratings', y='Discount Percentage', size="Sale Price", trendline="ols", title="relationship between discount price and number of ratings")
fig.show()
```

There is a positive relationship with discount percentage and number of ratings. Higher the disocunt, higher the number of ratings.
